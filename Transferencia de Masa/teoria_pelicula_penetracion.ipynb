# PARTE 1 · TEORÍA

## 1. Fundamentos de la transferencia de masa interfásica

### 1.1 ¿Por qué necesitamos coeficientes de transferencia?

En absorción, desorción, extracción, humidificación o evaporación, un componente **A** pasa de una fase a otra a través de una **interfase**. Dentro de cada fase, A se mueve por dos mecanismos:

- **Difusión molecular:** movimiento aleatorio de las moléculas debido a un gradiente de concentración (ley de Fick).
- **Transporte convectivo:** arrastre de A por el movimiento macroscópico del fluido (flujo laminar o turbulento).

Modelar la hidrodinámica completa cerca de una interfase es muy complejo. La ingeniería resuelve esto agrupando todo el efecto del flujo en un **coeficiente de transferencia de masa**, que relaciona el flux con una fuerza impulsora medible:

$$
\text{flux} = (\text{coeficiente}) \times (\text{fuerza impulsora})
$$

La **teoría de la película** y la **teoría de la penetración** son dos modelos físicos que permiten **predecir** ese coeficiente a partir de propiedades como la difusividad.

### 1.2 Flux y ley de Fick

El **flux molar total** de A respecto a un marco fijo es la suma de una contribución difusiva y una convectiva (arrastre global):

$$
N_A = J_A + y_A\,(N_A + N_B), \qquad J_A = -c\,D_{AB}\,\frac{dy_A}{dz}
$$

donde:

| Símbolo | Significado | Unidades típicas |
|---|---|---|
| $N_A$ | flux molar total de A | mol/(m²·s) |
| $J_A$ | flux difusivo (ley de Fick) | mol/(m²·s) |
| $c$ | concentración molar total de la mezcla | mol/m³ |
| $D_{AB}$ | difusividad de A en B | m²/s |
| $y_A$ | fracción molar de A (en gas; $x_A$ en líquido) | — |

Para un **gas ideal**: $c = \dfrac{P}{RT}$.

Dos casos límite son fundamentales:

| Caso | Condición | Flux |
|---|---|---|
| **Contradifusión equimolar** | $N_B = -N_A$ | $N_A = J_A$ (no hay arrastre neto) |
| **A difunde a través de B estancado** | $N_B = 0$ | $N_A = \dfrac{J_A}{1-y_A}$ (hay arrastre) |

El segundo caso es el típico de la absorción: el soluto A se disuelve en el líquido, pero el gas inerte B no.

### 1.3 Definición de los coeficientes convectivos

Se define un coeficiente para cada **fuerza impulsora**. La interfase se identifica con el subíndice $i$ y el seno del fluido con el subíndice $b$ (en inglés, *bulk*).

**Fase gas**

$$
N_A = k_c\,(c_{A,b}-c_{A,i}) = k_G\,(p_{A,b}-p_{A,i}) = k_y\,(y_{A,b}-y_{A,i})
$$

**Fase líquida**

$$
N_A = k_L\,(c_{A,i}-c_{A,b}) = k_x\,(x_{A,i}-x_{A,b})
$$

| Coeficiente | Fuerza impulsora | Unidades |
|---|---|---|
| $k_c$, $k_L$ | concentración molar | m/s |
| $k_G$ | presión parcial | mol/(m²·s·Pa) |
| $k_y$, $k_x$ | fracción molar | mol/(m²·s) |

**Relaciones entre coeficientes** (gas ideal, solución diluida):

$$
k_y = c\,k_c = \frac{P}{RT}\,k_c = P\,k_G, \qquad k_x = c_{\text{líq}}\,k_L
$$

**Corrección por B estancado.** Los coeficientes anteriores corresponden a contradifusión equimolar. Cuando A difunde a través de B estancado, el arrastre convectivo aumenta el flux y se usa un coeficiente corregido:

$$
k_c' = \frac{k_c}{(1-y_A)_{lm}}, \qquad (1-y_A)_{lm} = \frac{(1-y_{A,i})-(1-y_{A,b})}{\ln\dfrac{1-y_{A,i}}{1-y_{A,b}}}
$$

Cuando $y_A$ es pequeña, $(1-y_A)_{lm}\approx 1$ y la corrección es despreciable.

## 2. Teoría de la película (1.2.1)

### 2.1 Antecedentes

Es el modelo más antiguo y simple. La idea se atribuye a **Nernst (1904)** y fue extendida a la absorción gas-líquido por **Whitman y Lewis (1923-1924)**.

### 2.2 Modelo físico

Cerca de una interfase, el fluido se mueve de manera turbulenta en el seno, pero el movimiento se amortigua al acercarse a la superficie. La teoría **simplifica** esta situación así:

- Junto a la interfase existe una **película estancada** (o en flujo laminar) de espesor $\delta$.
- **Toda la resistencia** a la transferencia se concentra en esa película.
- Fuera de la película, el fluido está perfectamente mezclado: su concentración es uniforme e igual a la del seno $c_{A,b}$.
- En la película, A solo se transporta por **difusión molecular**, en **estado estacionario**.

```
 Seno del fluido        Película estancada        Interfase
 (mezclado, c_A,b)      (difusión, espesor δ)     (c_A,i)
 ─────────────────┬─────────────────────────────┬──────────
                  z = 0                         z = δ
```

### 2.3 Suposiciones

1. Estado estacionario.
2. Difusión unidimensional en la dirección $z$ (perpendicular a la interfase).
3. Propiedades físicas ($D_{AB}$, $c$) constantes.
4. Película de espesor uniforme $\delta$, sin flujo convectivo dentro de ella (salvo el arrastre por B estancado).
5. Sin reacción química.

### 2.4 Deducción: contradifusión equimolar

Con $N_B=-N_A$, el flux total es solo difusivo:

$$
N_A = -D_{AB}\,\frac{dc_A}{dz}
$$

En estado estacionario $N_A$ es constante a lo largo de la película, así que se integra entre $z=0$ ($c_A=c_{A,b}$) y $z=\delta$ ($c_A=c_{A,i}$):

$$
N_A \int_0^{\delta} dz = -D_{AB}\int_{c_{A,b}}^{c_{A,i}} dc_A
\;\;\Longrightarrow\;\;
N_A = \frac{D_{AB}}{\delta}\,(c_{A,b}-c_{A,i})
$$

El perfil de concentración es **lineal**. Al comparar con la definición $N_A = k_c\,(c_{A,b}-c_{A,i})$ se obtiene el resultado central:

$$
\boxed{\,k_c = \frac{D_{AB}}{\delta}\,}
$$

### 2.5 Deducción: A a través de B estancado

Con $N_B = 0$:

$$
N_A = -c\,D_{AB}\frac{dy_A}{dz} + y_A N_A
\;\;\Longrightarrow\;\;
N_A\,(1-y_A) = -c\,D_{AB}\frac{dy_A}{dz}
$$

Se separan variables e integran entre $z=0$ ($y_{A,b}$) y $z=\delta$ ($y_{A,i}$):

$$
N_A\,\delta = c\,D_{AB}\,\ln\frac{1-y_{A,i}}{1-y_{A,b}}
\;\;\Longrightarrow\;\;
\boxed{\,N_A = \frac{c\,D_{AB}}{\delta}\,\ln\frac{1-y_{A,i}}{1-y_{A,b}}\,}
$$

Usando el promedio logarítmico $(1-y_A)_{lm}$ definido antes, la expresión es equivalente a:

$$
N_A = \frac{D_{AB}}{\delta\,(1-y_A)_{lm}}\,(c_{A,b}-c_{A,i}) = k_c'\,(c_{A,b}-c_{A,i})
$$

El perfil de concentración ya **no es lineal**; sigue la forma

$$
1-y_A(z) = (1-y_{A,b})\,\exp\!\left(\frac{N_A\,z}{c\,D_{AB}}\right)
$$

### 2.6 Consecuencias y limitaciones

| Aspecto | Predicción de la teoría de la película |
|---|---|
| Dependencia con la difusividad | $k \propto D_{AB}^{\,1}$ |
| Dependencia con la hidrodinámica | $\delta$ disminuye al aumentar la turbulencia, y $k$ aumenta |
| Parámetro desconocido | $\delta$ no se mide directamente; se obtiene a partir de $k$ experimental |

**Limitación principal.** Los experimentos muestran que $k$ varía con $D_{AB}$ elevada a una potencia entre aproximadamente **0.5 y 0.67**, no a la primera potencia. Por eso la teoría de la película no es predictiva por sí sola; su valor está en que es **simple**, da la **estructura de resistencias en serie** y funciona bien cuando la película realmente es delgada y el estado es estacionario.

## 3. Teoría de las dos películas (Whitman)

Cuando se transfiere masa **entre dos fases** (por ejemplo gas y líquido), Whitman propuso aplicar el modelo de película a cada lado de la interfase:

```
   GAS                      INTERFASE                  LÍQUIDO
 y_G (seno) ── película gas ── y_i  ⇌  x_i ── película líquida ── x_L (seno)
                               (equilibrio)
```

### 3.1 Suposiciones adicionales

1. En la **interfase no hay resistencia**: las composiciones $y_i$ y $x_i$ están en **equilibrio**, $y_i = f(x_i)$.
2. Las dos películas están en **serie**.
3. En estado estacionario, el flux es el mismo en ambos lados.

### 3.2 Planteamiento

$$
N_A = k_y\,(y_G - y_i) = k_x\,(x_i - x_L)
$$

Reordenando se obtiene la **recta de unión** (o de fuerza impulsora) que pasa por el punto $(x_L,\,y_G)$ y por el punto de interfase $(x_i,\,y_i)$:

$$
\frac{y_G - y_i}{x_L - x_i} = -\frac{k_x}{k_y}
$$

El punto de interfase es la **intersección** de esa recta con la curva de equilibrio.

### 3.3 Coeficientes globales

Para evitar calcular la interfase se definen **coeficientes globales**, que usan la composición que tendría una fase si estuviera en equilibrio con la otra. Con una curva de equilibrio lineal local $y^* = m\,x$:

$$
N_A = K_y\,(y_G - y^*) , \qquad y^* = m\,x_L
$$

$$
\boxed{\,\frac{1}{K_y} = \frac{1}{k_y} + \frac{m}{k_x}\,}
\qquad
\boxed{\,\frac{1}{K_x} = \frac{1}{m\,k_y} + \frac{1}{k_x}\,}
\qquad
K_x = m\,K_y
$$

Cada término es una **resistencia**:

| Término | Significado |
|---|---|
| $1/k_y$ | resistencia de la película **gas** |
| $m/k_x$ | resistencia de la película **líquida** (expresada en base $y$) |

### 3.4 ¿Quién controla?

$$
\%\text{ resistencia gas} = \frac{1/k_y}{1/K_y}\times 100, \qquad
\%\text{ resistencia líquido} = \frac{m/k_x}{1/K_y}\times 100
$$

| Tipo de sistema | Pendiente $m$ | Fase controlante |
|---|---|---|
| Gas **muy soluble** (por ejemplo NH₃ en agua) | $m$ pequeña | **gas** |
| Gas **poco soluble** (por ejemplo CO₂, O₂ en agua) | $m$ grande | **líquido** |

Esto tiene una consecuencia práctica: para mejorar la transferencia hay que **aumentar la turbulencia en la fase que controla**.

## 4. Teoría de la penetración (1.2.2)

### 4.1 Antecedentes

Propuesta por **Higbie (1935)** al estudiar la absorción de gas en líquido durante exposiciones muy cortas, por ejemplo burbujas que ascienden o películas que descienden. La teoría de la película no explicaba bien esos casos porque **el estado estacionario nunca se alcanza**.

### 4.2 Modelo físico

- Los **elementos de fluido** del seno son llevados a la interfase por los remolinos turbulentos.
- Cada elemento permanece en contacto con la otra fase un tiempo fijo $t_c$ (**tiempo de contacto** o de exposición).
- Durante ese tiempo, A **penetra por difusión molecular en régimen transitorio**.
- Pasado $t_c$, el elemento regresa al seno y es sustituido por otro fresco, con composición del seno.

```
   Interfase (z = 0)
   ────────────────────────────  c = c_i  (siempre)
   │  elemento de fluido       │
   │  A penetra con el tiempo  │  → profundidad de penetración crece como √(D·t)
   ↓ z                         │
   c = c_b  lejos de la interfase
```

### 4.3 Suposiciones

1. Cada elemento se comporta como un **medio semi-infinito** (A no alcanza el fondo del elemento durante $t_c$).
2. La transferencia en el elemento es por **difusión molecular pura** y **no estacionaria**.
3. La concentración en la interfase es constante, $c_i$.
4. **Todos** los elementos tienen el **mismo tiempo de contacto** $t_c$.
5. Propiedades constantes, sin reacción.

### 4.4 Planteamiento matemático

Segunda ley de Fick (difusión unidimensional transitoria):

$$
\frac{\partial c_A}{\partial t} = D_{AB}\,\frac{\partial^2 c_A}{\partial z^2}
$$

Condiciones (con $z$ medida desde la interfase hacia el interior del fluido):

| Condición | Expresión |
|---|---|
| Inicial | $c_A(z,0) = c_b$ |
| En la interfase | $c_A(0,t) = c_i$ |
| Lejos de la interfase | $c_A(\infty,t) = c_b$ |

### 4.5 Solución

Con la variable de similaridad $\eta = \dfrac{z}{2\sqrt{D_{AB}\,t}}$, la ecuación diferencial parcial se reduce a una ecuación ordinaria, cuya solución es la **función error**:

$$
\frac{c_A - c_i}{c_b - c_i} = \operatorname{erf}\!\left(\frac{z}{2\sqrt{D_{AB}\,t}}\right)
$$

**Flux instantáneo en la interfase.** Se deriva el perfil en $z=0$ (la derivada de erf en cero es $2/\sqrt{\pi}$):

$$
\left.\frac{\partial c_A}{\partial z}\right|_{z=0} = \frac{c_b-c_i}{\sqrt{\pi\,D_{AB}\,t}}
\;\;\Longrightarrow\;\;
N_A(t) = -D_{AB}\left.\frac{\partial c_A}{\partial z}\right|_{z=0}
= (c_i-c_b)\sqrt{\frac{D_{AB}}{\pi\,t}}
$$

El flux es **infinito en $t=0$** y decae como $t^{-1/2}$, porque el gradiente es enorme cuando el elemento fresco toca la interfase.

**Flux promedio en el tiempo de contacto.**

$$
\bar N_A = \frac{1}{t_c}\int_0^{t_c} N_A(t)\,dt
= \frac{(c_i-c_b)}{t_c}\sqrt{\frac{D_{AB}}{\pi}}\;\bigl[\,2\sqrt{t}\,\bigr]_0^{t_c}
= 2\,(c_i-c_b)\sqrt{\frac{D_{AB}}{\pi\,t_c}}
$$

Al comparar con $\bar N_A = k_L\,(c_i-c_b)$:

$$
\boxed{\,k_L = 2\sqrt{\frac{D_{AB}}{\pi\,t_c}}\,}
$$

**Moles absorbidos por unidad de área** durante un tiempo $t$:

$$
\frac{n_A}{A} = \int_0^t N_A\,dt = 2\,(c_i-c_b)\sqrt{\frac{D_{AB}\,t}{\pi}}
$$

### 4.6 Profundidad de penetración y validez

El perfil de erf alcanza el 99 % de su variación total cuando $\eta\approx 1.8$, es decir, a una profundidad

$$
\delta_p \approx 3.6\,\sqrt{D_{AB}\,t_c}
$$

La hipótesis de medio semi-infinito es válida si $\delta_p \ll$ espesor del elemento de fluido (por ejemplo, el espesor de la película líquida o el diámetro de la burbuja).

### 4.7 Estimación del tiempo de contacto

$t_c$ no se conoce de antemano y depende de la geometría y del flujo:

| Situación | Estimación de $t_c$ |
|---|---|
| Burbuja ascendente de diámetro $d_b$ y velocidad $u_b$ | $t_c \approx d_b/u_b$ |
| Película descendente de longitud $L$ y velocidad superficial $u_s$ | $t_c \approx L/u_s$ |
| Gota que cae una altura $H$ a velocidad $u$ | $t_c \approx H/u$ |

### 4.8 Consecuencias

| Aspecto | Predicción de la teoría de la penetración |
|---|---|
| Dependencia con la difusividad | $k \propto D_{AB}^{\,1/2}$ (más cercana a los datos experimentales) |
| Dependencia con el tiempo de contacto | $k \propto t_c^{-1/2}$: más turbulencia, menor $t_c$ y mayor $k$ |
| Parámetro desconocido | $t_c$ |
| Espesor equivalente de película | $\delta_{eq} = D_{AB}/k_L = \tfrac{1}{2}\sqrt{\pi\,D_{AB}\,t_c}$ |

La teoría de la penetración **antecede** a la **teoría de la renovación superficial (Danckwerts, 1951)**, que relaja la hipótesis de $t_c$ único y supone una distribución de edades de los elementos de superficie. Esa teoría es el tema 1.2.3.

## 5. Analogía con la transferencia de calor

La difusión de masa y la conducción de calor obedecen ecuaciones matemáticamente idénticas. Por eso ambos modelos se aplican directamente a la transferencia de calor con el **coeficiente convectivo $h$**, definido por $q = h\,(T_i - T_b)$.

| Magnitud | Masa | Calor |
|---|---|---|
| Propiedad de transporte | $D_{AB}$ (m²/s) | $\alpha = \dfrac{k_t}{\rho\,c_p}$ (m²/s) |
| Fuerza impulsora | $\Delta c_A$ | $\Delta T$ |
| Coeficiente | $k_c$ (m/s) | $h$ (W/(m²·K)) |
| **Película** | $k_c = D_{AB}/\delta$ | $h = k_t/\delta$ |
| **Penetración** | $k_L = 2\sqrt{D_{AB}/(\pi t_c)}$ | $h = 2\sqrt{\dfrac{k_t\,\rho\,c_p}{\pi\,t_c}} = \dfrac{2\,k_t}{\sqrt{\pi\,\alpha\,t_c}}$ |

Donde $k_t$ es la conductividad térmica, $\rho$ la densidad y $c_p$ el calor específico. La correspondencia se obtiene sustituyendo $D_{AB}\to\alpha$ en la ecuación de difusión transitoria y $c_A\to T$.

## 6. Comparación de las dos teorías

| Característica | Película | Penetración |
|---|---|---|
| Autor y año | Nernst (1904), Whitman (1923) | Higbie (1935) |
| Régimen | Estacionario | **No** estacionario |
| Perfil de concentración | Lineal (equimolar) | Función error |
| Parámetro del modelo | espesor $\delta$ | tiempo de contacto $t_c$ |
| Coeficiente | $k = D/\delta$ | $k = 2\sqrt{D/(\pi t_c)}$ |
| Dependencia con $D$ | $D^{1}$ | $D^{1/2}$ |
| Dependencia experimental típica | $D^{0.5\text{ a }0.67}$ | |
| Mejor para | películas delgadas, procesos continuos y estables | contactos breves, burbujas, gotas, películas descendentes |
| Limitación | exponente de $D$ demasiado alto | requiere estimar $t_c$ |

**Equivalencia práctica.** Los dos modelos dan el mismo $k$ si se elige $\delta = \tfrac{1}{2}\sqrt{\pi D\,t_c}$, o bien $t_c = \dfrac{4D}{\pi k^2}$. Por eso la teoría de la película, aunque físicamente menos realista, sigue siendo útil como lenguaje común de ingeniería.

## 7. Visualización de los modelos

### 7.1 Perfil de concentración en la película: equimolar vs. B estancado

El perfil equimolar es una recta. Con B estancado hay arrastre convectivo, el perfil se curva y el flux es mayor. Para que la curvatura se aprecie se usan composiciones altas ($y_{A,b}=0.60$, $y_{A,i}=0.05$); con las del Problema 1 ambos perfiles casi coinciden.

````python
# Datos ilustrativos (composiciones altas para que se note la curvatura;
# P, T, D y delta son los del Problema 1)
P, T = 101325.0, 298.15      # Pa, K
D, delta = 2.3e-5, 1.0e-3    # m²/s, m
yb, yi = 0.60, 0.05          # fracción molar en el seno y en la interfase
c = P / (R * T)              # mol/m³

# Fluxes
N_eq = D / delta * c * (yb - yi)
N_B  = D * c / delta * np.log((1 - yi) / (1 - yb))

z = np.linspace(0, delta, 200)
y_eq = yb + (yi - yb) * z / delta                  # perfil lineal
y_B  = 1 - (1 - yb) * np.exp(N_B * z / (c * D))    # perfil con B estancado

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(z * 1e3, y_eq, label=f"Equimolar (N = {N_eq:.4f} mol/m²·s)", lw=2)
ax.plot(z * 1e3, y_B, "--", label=f"B estancado (N = {N_B:.4f} mol/m²·s)", lw=2)
ax.axvline(delta * 1e3, color="gray", ls=":")
ax.annotate("interfase", (delta * 1e3, yi), textcoords="offset points", xytext=(-55, 8), color="gray")
ax.annotate("seno del gas", (0, yb), textcoords="offset points", xytext=(5, 6), color="gray")
ax.set_xlabel("Posición z dentro de la película (mm)")
ax.set_ylabel("Fracción molar de A, $y_A$")
ax.set_title("Teoría de la película: perfiles de concentración")
ax.legend()
plt.show()
````

### 7.2 Perfiles de penetración y flux instantáneo (Higbie)

Se grafican los perfiles de erf para varios tiempos y el decaimiento del flux, junto con su promedio en $t_c$. Los datos corresponden al CO₂ en agua del Problema 3.

````python
# Datos del Problema 3 (CO2 en agua, 25 °C)
D_L = 1.9e-9      # m²/s
ci, cb = 33.8, 0.0  # mol/m³ (saturación en la interfase, seno sin soluto)
tc = 0.02         # s

# --- Perfiles adimensionales para varios tiempos ---
z = np.linspace(0, 60e-6, 400)  # hasta 60 µm
fig, axs = plt.subplots(1, 2, figsize=(11, 4))

for t in [0.002, 0.005, 0.01, 0.02]:
    cA = ci + (cb - ci) * erf(z / (2 * np.sqrt(D_L * t)))
    axs[0].plot(z * 1e6, cA, label=f"t = {t*1e3:.0f} ms")
axs[0].set_xlabel("Profundidad z (µm)")
axs[0].set_ylabel("$c_A$ (mol/m³)")
axs[0].set_title("Penetración de A en el elemento de fluido")
axs[0].legend()

# --- Flux instantáneo y promedio ---
t = np.linspace(1e-4, tc, 400)
N_t = (ci - cb) * np.sqrt(D_L / (np.pi * t))
N_avg = 2 * (ci - cb) * np.sqrt(D_L / (np.pi * tc))

axs[1].plot(t * 1e3, N_t * 1e3, lw=2, label="Flux instantáneo $N_A(t)$")
axs[1].axhline(N_avg * 1e3, color="tab:red", ls="--", label=f"Flux promedio = {N_avg*1e3:.2f} mmol/m²·s")
axs[1].set_xlabel("Tiempo (ms)")
axs[1].set_ylabel("Flux (mmol/m²·s)")
axs[1].set_title("Decaimiento del flux con el tiempo")
axs[1].legend()
plt.tight_layout()
plt.show()
````

### 7.3 Dependencia del coeficiente con la difusividad

Las dos teorías predicen pendientes distintas en un gráfico log-log de $k$ contra $D_{AB}$. Con ambas fijadas en el mismo punto de referencia, la diferencia se ve de inmediato.

## Problema 1 · Teoría de la película en fase gas

**Enunciado.** Un soluto A se transfiere desde el seno de una corriente gaseosa hasta una interfase líquida. Se modela la fase gas con la teoría de la película. Calcule:

a) el coeficiente $k_c$ y el flux de A suponiendo **contradifusión equimolar**;

b) el flux de A suponiendo que A difunde a través de **B estancado**, y el coeficiente corregido $k_c'$;

c) en qué porcentaje aumenta el flux por el arrastre convectivo;

d) el espesor de película equivalente si un experimento mide $k_c = 0.010$ m/s.

**Datos**

| Variable | Valor |
|---|---|
| Temperatura, $T$ | 25 °C = 298.15 K |
| Presión total, $P$ | 101 325 Pa |
| Difusividad de A en el gas, $D_{AB}$ | $2.3\times10^{-5}$ m²/s |
| Espesor de película, $\delta$ | 1.0 mm |
| Fracción molar de A en el seno, $y_{A,b}$ | 0.10 |
| Fracción molar de A en la interfase, $y_{A,i}$ | 0.02 |

### Solución paso a paso

**Paso 1. Concentración molar total** (gas ideal):

$$
c = \frac{P}{RT} = \frac{101\,325}{8.314\times 298.15} = 40.88\ \text{mol/m}^3
$$

**Paso 2. Coeficiente con la teoría de la película:**

$$
k_c = \frac{D_{AB}}{\delta} = \frac{2.3\times10^{-5}}{1.0\times10^{-3}} = 0.0230\ \text{m/s}
$$

**Paso 3. Inciso (a), flux equimolar.** Con $k_y = c\,k_c$:

$$
N_A = k_c\,c\,(y_{A,b}-y_{A,i}) = 0.0230\times 40.88\times(0.10-0.02) = 0.0752\ \text{mol/(m}^2\text{·s)}
$$

**Paso 4. Inciso (b), flux con B estancado:**

$$
N_A = \frac{c\,D_{AB}}{\delta}\,\ln\frac{1-y_{A,i}}{1-y_{A,b}}
= 0.0230\times 40.88\times\ln\frac{0.98}{0.90}
= 0.9402\times 0.08516 = 0.0801\ \text{mol/(m}^2\text{·s)}
$$

**Paso 5. Coeficiente corregido.** Primero el promedio logarítmico:

$$
(1-y_A)_{lm} = \frac{0.98-0.90}{\ln(0.98/0.90)} = 0.9394
\;\;\Rightarrow\;\;
k_c' = \frac{k_c}{(1-y_A)_{lm}} = \frac{0.0230}{0.9394} = 0.0245\ \text{m/s}
$$

**Paso 6. Inciso (c), aumento del flux:**

$$
\frac{N_{A,\text{B estancado}}}{N_{A,\text{equimolar}}} = \frac{0.0801}{0.0752} = 1.064
\;\;\Rightarrow\;\; +6.4\ \%
$$

**Paso 7. Inciso (d), espesor equivalente:**

$$
\delta = \frac{D_{AB}}{k_c} = \frac{2.3\times10^{-5}}{0.010} = 2.3\times10^{-3}\ \text{m} = 2.3\ \text{mm}
$$

### Verificación con Python

````python
# --- Datos ---
T, P = 298.15, 101325.0
D, delta = 2.3e-5, 1.0e-3
yb, yi = 0.10, 0.02

# Paso 1
c = P / (R * T)
# Paso 2
kc = D / delta
# Paso 3: equimolar
N_eq = kc * c * (yb - yi)
# Paso 4: B estancado
N_B = (c * D / delta) * np.log((1 - yi) / (1 - yb))
# Paso 5: promedio logarítmico y coeficiente corregido
y_lm = ((1 - yi) - (1 - yb)) / np.log((1 - yi) / (1 - yb))
kc_corr = kc / y_lm
# Paso 6
aumento = (N_B / N_eq - 1) * 100
# Paso 7
kc_exp = 0.010
delta_eq = D / kc_exp

print(f"c                       = {c:.2f} mol/m³")
print(f"k_c                     = {kc:.4f} m/s")
print(f"N_A (equimolar)         = {N_eq:.4f} mol/(m²·s)")
print(f"N_A (B estancado)       = {N_B:.4f} mol/(m²·s)")
print(f"(1-y)_lm                = {y_lm:.4f}")
print(f"k_c' (corregido)        = {kc_corr:.4f} m/s")
print(f"Aumento del flux        = {aumento:.1f} %")
print(f"δ equivalente (k=0.010) = {delta_eq*1e3:.2f} mm")
````

### Análisis

- Con $y_A$ del orden de 0.1 el arrastre convectivo aumenta el flux apenas un 6 %. Si la fracción molar de A fuera mucho mayor, la corrección crecería rápidamente y no podría ignorarse.
- Un coeficiente medido menor ($0.010$ m/s) implica una película equivalente **más gruesa** (2.3 mm): menos turbulencia, mayor resistencia.

## Problema 2 · Dos películas, coeficientes globales y fase controlante

**Enunciado.** En una sección de una columna de absorción, un soluto A pasa de una mezcla gaseosa a agua. Para las condiciones locales dadas, calcule en dos casos:

a) las composiciones en la interfase $(x_i,\,y_i)$;

b) el coeficiente global $K_y$ y el flux de A;

c) el porcentaje de la resistencia total que corresponde a cada fase.

Caso **A**: gas muy soluble, $m = 0.5$. Caso **B**: gas poco soluble, $m = 50$. En ambos, el equilibrio local es $y^* = m\,x$.

**Datos**

| Variable | Valor |
|---|---|
| Composición del gas en el seno, $y_G$ | 0.05 |
| Composición del líquido en el seno, $x_L$ | $2\times10^{-4}$ |
| $k_y$ | $1.5\times10^{-3}$ kmol/(m²·s) |
| $k_x$ | $6.0\times10^{-3}$ kmol/(m²·s) |

### Solución paso a paso

**Paso 1. Ecuaciones.** En la interfase se cumple el equilibrio y la igualdad de fluxes:

$$
k_y\,(y_G-y_i) = k_x\,(x_i-x_L), \qquad y_i = m\,x_i
$$

Despejando $x_i$:

$$
x_i = \frac{k_y\,y_G + k_x\,x_L}{k_y\,m + k_x}, \qquad y_i = m\,x_i
$$

**Paso 2. Interfase, caso A ($m=0.5$):**

$$
x_i = \frac{(1.5\times10^{-3})(0.05)+(6.0\times10^{-3})(2\times10^{-4})}{(1.5\times10^{-3})(0.5)+6.0\times10^{-3}}
= \frac{7.62\times10^{-5}}{6.75\times10^{-3}} = 0.01129
$$

$$
y_i = 0.5\times 0.01129 = 0.00564
$$

**Paso 3. Interfase, caso B ($m=50$):**

$$
x_i = \frac{7.62\times10^{-5}}{(1.5\times10^{-3})(50)+6.0\times10^{-3}}
= \frac{7.62\times10^{-5}}{0.081} = 9.41\times10^{-4},
\qquad y_i = 50\times 9.41\times10^{-4} = 0.0470
$$

**Paso 4. Coeficiente global:**

$$
\frac{1}{K_y} = \frac{1}{k_y}+\frac{m}{k_x}
$$

- Caso A: $\dfrac{1}{K_y} = 666.7 + 83.3 = 750$, entonces $K_y = 1.333\times10^{-3}$ kmol/(m²·s).
- Caso B: $\dfrac{1}{K_y} = 666.7 + 8333.3 = 9000$, entonces $K_y = 1.111\times10^{-4}$ kmol/(m²·s).

**Paso 5. Flux** con $y^* = m\,x_L$:

- Caso A: $y^* = 0.0001$, entonces $N_A = 1.333\times10^{-3}\,(0.05-0.0001) = 6.65\times10^{-5}$ kmol/(m²·s).
- Caso B: $y^* = 0.01$, entonces $N_A = 1.111\times10^{-4}\,(0.05-0.01) = 4.44\times10^{-6}$ kmol/(m²·s).

**Paso 6. Porcentaje de resistencia:**

| Caso | Resistencia gas $1/k_y$ | Resistencia líquido $m/k_x$ | Controla |
|---|---|---|---|
| A ($m=0.5$) | $666.7/750 = 88.9\ \%$ | $11.1\ \%$ | **Gas** |
| B ($m=50$) | $666.7/9000 = 7.4\ \%$ | $92.6\ \%$ | **Líquido** |

### Verificación con Python

````python
ky, kx = 1.5e-3, 6.0e-3      # kmol/(m²·s)
yG, xL = 0.05, 2.0e-4

def dos_peliculas(m):
    # Interfase
    xi = (ky * yG + kx * xL) / (ky * m + kx)
    yi = m * xi
    # Global
    r_gas, r_liq = 1 / ky, m / kx
    Ky = 1 / (r_gas + r_liq)
    ystar = m * xL
    N = Ky * (yG - ystar)
    return dict(m=m, xi=xi, yi=yi, Ky=Ky, N=N,
                pct_gas=100 * r_gas / (r_gas + r_liq),
                pct_liq=100 * r_liq / (r_gas + r_liq))

res = {nombre: dos_peliculas(m) for nombre, m in [("A (m=0.5)", 0.5), ("B (m=50)", 50.0)]}

for nombre, r in res.items():
    print(f"Caso {nombre}")
    print(f"  x_i = {r['xi']:.5f}   y_i = {r['yi']:.5f}")
    print(f"  K_y = {r['Ky']:.4e} kmol/(m²·s)   N_A = {r['N']:.3e} kmol/(m²·s)")
    print(f"  Resistencia gas = {r['pct_gas']:.1f} %   líquido = {r['pct_liq']:.1f} %")
    # Comprobación de igualdad de fluxes
    N_gas = ky * (yG - r['yi']); N_liq = kx * (r['xi'] - xL)
    print(f"  Comprobación: k_y(yG-yi) = {N_gas:.3e}   k_x(xi-xL) = {N_liq:.3e}\n")
````

````python
# Representación gráfica de la recta de unión
fig, axs = plt.subplots(1, 2, figsize=(11, 4.8), sharey=False)
for ax, (nombre, r) in zip(axs, res.items()):
    m = r["m"]
    x = np.linspace(0, max(r["xi"], xL) * 1.6, 100)
    ax.plot(x, m * x, "k", lw=2, label="Equilibrio $y^* = m\\,x$")
    ax.plot([xL, r["xi"]], [yG, r["yi"]], "r-o", label="Recta de unión (pendiente $-k_x/k_y$)")
    ax.plot(xL, yG, "bs", ms=8, label="Seno $(x_L, y_G)$")
    ax.plot(r["xi"], r["yi"], "g^", ms=9, label="Interfase $(x_i, y_i)$")
    ax.set_xlabel("x (fracción molar en líquido)")
    ax.set_ylabel("y (fracción molar en gas)")
    ax.set_title(f"Caso {nombre}")
    ax.legend(fontsize=8, loc="upper center", bbox_to_anchor=(0.5, -0.18), ncol=2, frameon=False)
plt.tight_layout()
plt.show()
````

### Análisis

- En el caso **A** (gas muy soluble) casi toda la caída de composición ocurre en el gas: la interfase queda cerca del equilibrio con el líquido. Para mejorar la transferencia conviene aumentar la turbulencia del **gas**.
- En el caso **B** (gas poco soluble) la mayor parte de la resistencia está en el líquido y $y_i\approx y_G$. Conviene aumentar la turbulencia del **líquido**.
- En ambos casos $k_y(y_G-y_i) = k_x(x_i-x_L)$, lo que confirma que los fluxes de las dos películas coinciden.

## Problema 3 · Teoría de la penetración en una burbuja

**Enunciado.** Una burbuja de CO₂ puro asciende por agua a 25 °C. Calcule con la teoría de la penetración:

a) el tiempo de contacto;

b) el coeficiente de transferencia de masa en la fase líquida $k_L$;

c) el flux promedio de CO₂ hacia el agua, suponiendo que el agua no contiene CO₂ disuelto;

d) el espesor de película equivalente y la profundidad de penetración, para verificar la hipótesis de medio semi-infinito;

e) los moles absorbidos por una burbuja durante su exposición.

**Datos**

| Variable | Valor |
|---|---|
| Diámetro de la burbuja, $d_b$ | 5.0 mm |
| Velocidad de ascenso, $u_b$ | 0.25 m/s |
| Difusividad del CO₂ en agua, $D_{AB}$ | $1.9\times10^{-9}$ m²/s |
| Concentración de saturación en la interfase, $c_i$ | 33.8 mol/m³ |
| Concentración en el seno del líquido, $c_b$ | 0 |

### Solución paso a paso

**Paso 1. Tiempo de contacto.** Cada elemento de líquido está expuesto mientras la burbuja pasa una distancia igual a su diámetro:

$$
t_c = \frac{d_b}{u_b} = \frac{0.005}{0.25} = 0.020\ \text{s}
$$

**Paso 2. Coeficiente de Higbie:**

$$
k_L = 2\sqrt{\frac{D_{AB}}{\pi\,t_c}} = 2\sqrt{\frac{1.9\times10^{-9}}{\pi\,(0.020)}} = 3.48\times10^{-4}\ \text{m/s}
$$

**Paso 3. Flux promedio:**

$$
\bar N_A = k_L\,(c_i-c_b) = (3.48\times10^{-4})(33.8-0) = 1.18\times10^{-2}\ \text{mol/(m}^2\text{·s)}
$$

**Paso 4. Verificación del modelo.**

$$
\delta_{eq} = \frac{D_{AB}}{k_L} = \frac{1.9\times10^{-9}}{3.48\times10^{-4}} = 5.5\ \mu\text{m}
$$

$$
\delta_p \approx 3.6\sqrt{D_{AB}\,t_c} = 3.6\sqrt{(1.9\times10^{-9})(0.020)} = 22\ \mu\text{m}
$$

Como $22\ \mu\text{m}\ll 5\ \text{mm}$, la hipótesis de medio semi-infinito es válida.

**Paso 5. Moles absorbidos.** Por unidad de área durante $t_c$:

$$
\frac{n_A}{A} = 2\,c_i\sqrt{\frac{D_{AB}\,t_c}{\pi}} = 2.35\times10^{-4}\ \text{mol/m}^2
$$

El área de la burbuja es $A = \pi d_b^2 = 7.85\times10^{-5}\ \text{m}^2$, así que

$$
n_A = (2.35\times10^{-4})(7.85\times10^{-5}) = 1.85\times10^{-8}\ \text{mol}
$$

### Verificación con Python

````python
D = 1.9e-9      # m²/s
d_b, u_b = 5.0e-3, 0.25
ci, cb = 33.8, 0.0

tc = d_b / u_b                                   # Paso 1
kL = 2 * np.sqrt(D / (np.pi * tc))               # Paso 2
N_prom = kL * (ci - cb)                          # Paso 3
delta_eq = D / kL                                # Paso 4
delta_p = 3.6 * np.sqrt(D * tc)
n_por_area = 2 * (ci - cb) * np.sqrt(D * tc / np.pi)   # Paso 5
A_burbuja = np.pi * d_b**2
n_burbuja = n_por_area * A_burbuja

print(f"t_c                  = {tc:.3f} s")
print(f"k_L                  = {kL:.3e} m/s")
print(f"Flux promedio        = {N_prom:.3e} mol/(m²·s)")
print(f"δ equivalente        = {delta_eq*1e6:.2f} µm")
print(f"Profundidad penetr.  = {delta_p*1e6:.1f} µm  (≪ {d_b*1e3:.0f} mm → semi-infinito válido)")
print(f"n por área           = {n_por_area:.3e} mol/m²")
print(f"n por burbuja        = {n_burbuja:.3e} mol")
````

### Análisis

- El coeficiente obtenido, del orden de $10^{-4}$ m/s, es típico de la fase líquida en contactos breves.
- Si la burbuja fuera más pequeña o subiera más rápido, $t_c$ disminuiría y $k_L$ aumentaría como $t_c^{-1/2}$.
- El flux **promedio** es exactamente el doble del flux instantáneo en $t = t_c$, resultado propio del decaimiento $t^{-1/2}$.

## Problema 4 · Película descendente (columna de pared mojada)

**Enunciado.** En una columna de pared mojada, una película de agua desciende por la pared interior de un tubo en contacto con CO₂ puro a 25 °C. Calcule con la teoría de la penetración:

a) el tiempo de contacto y el coeficiente $k_L$;

b) el flux promedio y la velocidad total de absorción de CO₂;

c) si la hipótesis de medio semi-infinito es válida para una película de 1.0 mm de espesor.

**Datos**

| Variable | Valor |
|---|---|
| Longitud del tubo, $L$ | 1.5 m |
| Diámetro interior, $d$ | 25 mm |
| Velocidad en la superficie de la película, $u_s$ | 0.50 m/s |
| Espesor de la película | 1.0 mm |
| $D_{AB}$ del CO₂ en agua | $1.9\times10^{-9}$ m²/s |
| $c_i$ (saturación) | 33.8 mol/m³ |
| $c_b$ | 0 |

### Solución paso a paso

**Paso 1. Tiempo de contacto.** Un elemento de superficie recorre toda la longitud del tubo:

$$
t_c = \frac{L}{u_s} = \frac{1.5}{0.50} = 3.0\ \text{s}
$$

**Paso 2. Coeficiente:**

$$
k_L = 2\sqrt{\frac{D_{AB}}{\pi\,t_c}} = 2\sqrt{\frac{1.9\times10^{-9}}{\pi\,(3.0)}} = 2.84\times10^{-5}\ \text{m/s}
$$

**Paso 3. Flux promedio:**

$$
\bar N_A = k_L\,c_i = (2.84\times10^{-5})(33.8) = 9.60\times10^{-4}\ \text{mol/(m}^2\text{·s)}
$$

**Paso 4. Velocidad total de absorción.** El área interfacial es $A = \pi\,d\,L = \pi(0.025)(1.5) = 0.1178\ \text{m}^2$:

$$
\dot n_A = \bar N_A\,A = (9.60\times10^{-4})(0.1178) = 1.13\times10^{-4}\ \text{mol/s} \;(= 0.113\ \text{mmol/s})
$$

**Paso 5. Validez del modelo:**

$$
\delta_p \approx 3.6\sqrt{D_{AB}\,t_c} = 3.6\sqrt{(1.9\times10^{-9})(3.0)} = 0.27\ \text{mm}
$$

Como $0.27\ \text{mm} < 1.0\ \text{mm}$, el CO₂ no alcanza la pared y el modelo de medio semi-infinito es aceptable. Si el tubo fuera mucho más largo (mayor $t_c$), $\delta_p$ se acercaría al espesor de la película y habría que usar otro modelo.

### Verificación con Python

````python
D = 1.9e-9
L, d_tubo, u_s = 1.5, 0.025, 0.50
espesor = 1.0e-3
ci = 33.8

tc = L / u_s
kL = 2 * np.sqrt(D / (np.pi * tc))
N_prom = kL * ci
A = np.pi * d_tubo * L
n_dot = N_prom * A
delta_p = 3.6 * np.sqrt(D * tc)

print(f"t_c            = {tc:.1f} s")
print(f"k_L            = {kL:.3e} m/s")
print(f"Flux promedio  = {N_prom:.3e} mol/(m²·s)")
print(f"Área           = {A:.4f} m²")
print(f"Absorción      = {n_dot:.3e} mol/s  ({n_dot*1e3:.3f} mmol/s)")
print(f"δ_p            = {delta_p*1e3:.2f} mm  →  "
      f"{'válido' if delta_p < espesor else 'NO válido'} frente a {espesor*1e3:.1f} mm")

# Sensibilidad: k_L vs longitud del tubo
Ls = np.linspace(0.2, 5, 100)
kLs = 2 * np.sqrt(D / (np.pi * Ls / u_s))
plt.figure(figsize=(6.2, 3.8))
plt.plot(Ls, kLs * 1e5, lw=2)
plt.axvline(L, color="gray", ls=":")
plt.xlabel("Longitud del tubo, L (m)")
plt.ylabel("$k_L$ (×10⁻⁵ m/s)")
plt.title("El coeficiente cae con la longitud: $k_L \\propto L^{-1/2}$")
plt.show()
````

### Análisis

- El coeficiente es menor que en el problema de la burbuja porque el tiempo de contacto es 150 veces mayor ($3.0$ s frente a $0.02$ s), y $k_L\propto t_c^{-1/2}$.
- El coeficiente promedio **disminuye** si el tubo se alarga: los elementos de la parte baja llevan más tiempo expuestos, el gradiente en su superficie es menor y aportan menos flux que los de la parte alta.

## Problema 5 · Coeficiente de calor con película y penetración

**Enunciado.** Un paquete de agua líquida entra en contacto breve con una superficie y se calienta por conducción transitoria. Calcule el coeficiente convectivo $h$:

a) con la teoría de la **penetración**, para un tiempo de contacto de 0.020 s;

b) el espesor de película $\delta_{eq}$ con el que la teoría de la película daría el mismo $h$;

c) la profundidad térmica de penetración.

**Datos** (agua a 25 °C)

| Variable | Valor |
|---|---|
| Conductividad térmica, $k_t$ | 0.60 W/(m·K) |
| Densidad, $\rho$ | 997 kg/m³ |
| Calor específico, $c_p$ | 4180 J/(kg·K) |
| Tiempo de contacto, $t_c$ | 0.020 s |

### Solución paso a paso

**Paso 1. Difusividad térmica:**

$$
\alpha = \frac{k_t}{\rho\,c_p} = \frac{0.60}{(997)(4180)} = 1.44\times10^{-7}\ \text{m}^2/\text{s}
$$

**Paso 2. Coeficiente (penetración):**

$$
h = 2\sqrt{\frac{k_t\,\rho\,c_p}{\pi\,t_c}} = 2\sqrt{\frac{(0.60)(997)(4180)}{\pi\,(0.020)}} = 1.26\times10^{4}\ \text{W/(m}^2\text{·K)}
$$

**Paso 3. Espesor equivalente.** De $h = k_t/\delta_{eq}$:

$$
\delta_{eq} = \frac{k_t}{h} = \frac{0.60}{1.26\times10^{4}} = 47.6\ \mu\text{m}
$$

**Paso 4. Profundidad térmica de penetración:**

$$
\delta_p \approx 3.6\sqrt{\alpha\,t_c} = 3.6\sqrt{(1.44\times10^{-7})(0.020)} = 193\ \mu\text{m}
$$

### Verificación con Python

````python
kt, rho, cp, tc = 0.60, 997.0, 4180.0, 0.020

alpha = kt / (rho * cp)
h = 2 * np.sqrt(kt * rho * cp / (np.pi * tc))
delta_eq = kt / h
delta_p = 3.6 * np.sqrt(alpha * tc)

print(f"α          = {alpha:.3e} m²/s")
print(f"h          = {h:.0f} W/(m²·K)")
print(f"δ_eq       = {delta_eq*1e6:.1f} µm")
print(f"δ_p        = {delta_p*1e6:.0f} µm")
````

### Análisis

- Un contacto muy breve produce coeficientes de calor altos (del orden de $10^4$ W/(m²·K)). Por eso las superficies renovadas con frecuencia transfieren calor con tanta eficiencia.
- La analogía es directa: bastó sustituir $D_{AB}\to\alpha$ y $k\to h/(\rho c_p)$ en el resultado de masa.

## Problema 6 · ¿Qué teoría describe mejor los datos?

**Enunciado.** En el mismo equipo y bajo las mismas condiciones hidrodinámicas, se mide el coeficiente de transferencia de masa de dos solutos con difusividades distintas:

| Soluto | $D_{AB}$ (m²/s) | $k_L$ medido (m/s) |
|---|---|---|
| 1 | $1.0\times10^{-9}$ | $2.0\times10^{-4}$ |
| 2 | $2.5\times10^{-9}$ | $3.4\times10^{-4}$ |

a) Determine el exponente $n$ en $k_L\propto D_{AB}^{\,n}$.

b) Compare con las predicciones de las teorías de la película y de la penetración.

c) Calcule el espesor de película y el tiempo de contacto equivalentes para cada soluto.

### Solución paso a paso

**Paso 1. Exponente experimental.** De $k_2/k_1 = (D_2/D_1)^n$:

$$
n = \frac{\ln(k_2/k_1)}{\ln(D_2/D_1)} = \frac{\ln(3.4/2.0)}{\ln(2.5)} = \frac{0.5306}{0.9163} = 0.58
$$

**Paso 2. Predicciones de cada teoría para el soluto 2** (partiendo del soluto 1):

| Teoría | Cálculo | $k_2$ predicho |
|---|---|---|
| Película ($n=1$) | $2.0\times10^{-4}\times 2.5$ | $5.0\times10^{-4}$ m/s |
| Penetración ($n=0.5$) | $2.0\times10^{-4}\times\sqrt{2.5}$ | $3.2\times10^{-4}$ m/s |
| **Medido** | | $3.4\times10^{-4}$ m/s |

**Paso 3. Espesor y tiempo de contacto equivalentes:**

$$
\delta = \frac{D}{k}, \qquad t_c = \frac{4\,D}{\pi\,k^{2}}
$$

| Soluto | $\delta$ ($\mu$m) | $t_c$ (s) |
|---|---|---|
| 1 | 5.0 | 0.0318 |
| 2 | 7.4 | 0.0275 |

### Verificación con Python

````python
D1, k1 = 1.0e-9, 2.0e-4
D2, k2 = 2.5e-9, 3.4e-4

n = np.log(k2 / k1) / np.log(D2 / D1)
k2_pel = k1 * (D2 / D1) ** 1.0
k2_pen = k1 * (D2 / D1) ** 0.5

print(f"Exponente experimental n = {n:.3f}")
print(f"k2 película   (n=1.0) = {k2_pel:.2e} m/s")
print(f"k2 penetración(n=0.5) = {k2_pen:.2e} m/s")
print(f"k2 medido             = {k2:.2e} m/s")
print()
for nombre, D, k in [("1", D1, k1), ("2", D2, k2)]:
    delta = D / k
    tc = 4 * D / (np.pi * k**2)
    print(f"Soluto {nombre}: δ = {delta*1e6:.1f} µm   t_c = {tc:.4f} s")

# Gráfica log-log con la recta de cada teoría
Dg = np.logspace(-9.2, -8.3, 50)
plt.figure(figsize=(6.5, 4.2))
plt.loglog(Dg, k1 * (Dg / D1) ** 1.0, label="Película ($n=1$)")
plt.loglog(Dg, k1 * (Dg / D1) ** 0.5, label="Penetración ($n=0.5$)")
plt.loglog(Dg, k1 * (Dg / D1) ** n, "k--", label=f"Ajuste experimental ($n={n:.2f}$)")
plt.plot([D1, D2], [k1, k2], "ko", ms=8, label="Datos medidos")
plt.xlabel("$D_{AB}$ (m²/s)")
plt.ylabel("$k_L$ (m/s)")
plt.legend()
plt.title("Comparación con los datos experimentales")
plt.show()
````

### Análisis

- El exponente medido ($n\approx 0.58$) cae dentro del intervalo experimental típico (0.5 a 0.67) y queda **mucho más cerca de la penetración** que de la película.
- Esto no invalida la teoría de la película como herramienta de diseño, pero sí advierte que su interpretación física ($k=D/\delta$ con $\delta$ constante) no se sostiene: $\delta$ **depende** del soluto (5.0 frente a 7.4 µm).
- Los $t_c$ equivalentes de los dos solutos son parecidos (0.032 s y 0.028 s), como se espera si el mecanismo hidrodinámico es el mismo.

# PARTE 4 · RESUMEN Y REFERENCIAS

## Resumen de fórmulas

| Concepto | Expresión |
|---|---|
| Gas ideal | $c = P/(RT)$ |
| Conversión de coeficientes | $k_y = c\,k_c = P\,k_G$, $\;k_x = c_{\text{líq}}\,k_L$ |
| **Película**, coeficiente | $k_c = D_{AB}/\delta$ |
| **Película**, B estancado | $N_A = \dfrac{c\,D_{AB}}{\delta}\ln\dfrac{1-y_{A,i}}{1-y_{A,b}}$ |
| Corrección por B estancado | $k_c' = k_c/(1-y_A)_{lm}$ |
| Dos películas | $\dfrac{1}{K_y} = \dfrac{1}{k_y}+\dfrac{m}{k_x}$, $\;\dfrac{1}{K_x} = \dfrac{1}{m\,k_y}+\dfrac{1}{k_x}$ |
| Interfase (dos películas) | pendiente de la recta de unión $= -k_x/k_y$ |
| **Penetración**, perfil | $\dfrac{c_A-c_i}{c_b-c_i} = \operatorname{erf}\!\Bigl(\dfrac{z}{2\sqrt{D\,t}}\Bigr)$ |
| Penetración, flux instantáneo | $N_A(t) = (c_i-c_b)\sqrt{D/(\pi t)}$ |
| Penetración, coeficiente | $k_L = 2\sqrt{D_{AB}/(\pi\,t_c)}$ |
| Moles por área | $n_A/A = 2\,(c_i-c_b)\sqrt{D\,t/\pi}$ |
| Profundidad de penetración | $\delta_p \approx 3.6\sqrt{D\,t_c}$ |
| Equivalencia entre teorías | $\delta_{eq} = D/k$, $\;t_c = 4D/(\pi k^2)$ |
| **Calor**, película | $h = k_t/\delta$ |
| **Calor**, penetración | $h = 2\sqrt{k_t\,\rho\,c_p/(\pi\,t_c)}$ |

## Ideas clave

1. Un coeficiente de transferencia **condensa** la hidrodinámica en un solo número; la teoría de la película y la de la penetración son modelos que lo estiman a partir de $D_{AB}$.
2. En película: $k\propto D$ y el parámetro es $\delta$. En penetración: $k\propto D^{1/2}$ y el parámetro es $t_c$.
3. Con dos fases, las resistencias se suman en serie y **la fase de mayor resistencia controla**: aumentar la turbulencia donde no controla casi no ayuda.
4. La analogía masa-calor permite usar los mismos modelos cambiando $D_{AB}$ por la difusividad térmica $\alpha$.

## Siguiente tema

**1.2.3 Teoría de la renovación superficial (Danckwerts):** generaliza la penetración permitiendo que los elementos tengan una distribución de tiempos de contacto, con $k_L=\sqrt{D_{AB}\,s}$, donde $s$ es la fracción de superficie renovada por unidad de tiempo.
